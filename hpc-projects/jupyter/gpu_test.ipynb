```python
import time, torch
assert torch.cuda.is_available(), "CUDA not active!"
print("Use GPU:", torch.cuda.get_device_name(0))
d = torch.device("cuda:0")
a = torch.randn((12288, 12288), device=d, dtype=torch.float16)
b = torch.randn((12288, 12288), device=d, dtype=torch.float16)
print("Start 30 seconds GPU full load...")
t0 = time.time()
while time.time() - t0 < 30:
    c = torch.matmul(a, b)
    torch.cuda.synchronize()
print("30s CUDA full load finished!")
```
